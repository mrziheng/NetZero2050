```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

```python
soc_status = pd.read_pickle('./data/storage/soc_status.pkl')

median_ = soc_status['median']
mean_ = soc_status['mean']
q_25 = soc_status['q_25']
q_75 = soc_status['q_75']
q_0 = soc_status['q_0']
q_100 = soc_status['q_100']
```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 20

days = range(365)

fig,ax = plt.subplots(figsize=(24,5),dpi=600)

# 颜色映射
colors = "#2082af"

# 绘制50%区间
ax.fill_between(days, q_25, q_75, color=colors, alpha=0.3)
# 绘制100%区间
ax.fill_between(days, q_0, q_100, color=colors, alpha=0.15)

ax.plot(days, median_, color=colors, linewidth=1.5)

ax.axhline(0.35,color='purple',ls=':')
ax.axhline(0.65,color='purple',ls=':')

ax.set_ylabel('State of charge (daily mean, 0~1)',font={"size":20})

ax.set_xticks(range(0,365,15),range(0,365,15))

ax.set_xlabel('Day of year')

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=6, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=3, 
               color='blue')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_ylim(0,1)
ax.set_xlim(0,365)

legend_ax = fig.add_axes([0.15, 0.62, 0.05, 0.28])

legend_ax.axis('off')

legend_ax.spines['top'].set_visible(False)
legend_ax.spines['right'].set_visible(False)
legend_ax.spines['bottom'].set_visible(False)
legend_ax.spines['left'].set_visible(False)

legend_ax.set_xticks([],[])
legend_ax.set_yticks([],[])

# 定义图例参数
sources = ['MC']
x_positions = [0]

box_width = 0.5
box_height = 0.8

bottom = 0.1

# 绘制图例
for i, (source, x_pos) in enumerate(zip(sources, x_positions)):
    # 绘制100%框
    legend_ax.add_patch(Rectangle((x_pos, bottom), 
                                  box_width, 
                                  box_height,
                                  facecolor=colors,
                                  edgecolor='black', 
                                  alpha=0.15))
    
    legend_ax.add_patch(Rectangle((x_pos, bottom), 
                                  box_width, 
                                  box_height,
                                  facecolor='none',
                                  edgecolor='black'))
    
    # 绘制50%框
    legend_ax.add_patch(Rectangle((x_pos, bottom+0.25*box_height), 
                                  box_width, 
                                  box_height/2,
                                  facecolor=colors, 
                                  alpha=0.3))
    
    # 绘制中位数线
    legend_ax.plot([x_pos, x_pos + box_width], 
                   [bottom+0.5*box_height, 
                    bottom+0.5*box_height], 
                    color=colors, 
                    linewidth=1.5)
    
    if source == sources[-1]:
        legend_ax.text(x_pos + box_width + 0.025, bottom+0.5*box_height, 'Median', va='center',fontsize=18)
        legend_ax.text(x_pos + box_width + 0.025, bottom+box_height, '100%', va='center',fontsize=18)
        legend_ax.text(x_pos + box_width + 0.025, bottom+0.75*box_height, '75%', va='center',fontsize=18)

# 设置图例坐标轴的范围
legend_ax.set_xlim(-0.1, 1.1)
legend_ax.set_ylim(0.0, 1.0)


plt.savefig('./fig/sto_dischar.png',bbox_inches='tight')
plt.savefig('./fig/sto_dischar.pdf',bbox_inches='tight')

plt.close()
```
