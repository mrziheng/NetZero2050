```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Rectangle
import seaborn as sns
```

```python
vre_profit = pd.read_csv('./data/vre_return/vre_prifit.csv')
```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 20

color = {
    'onshore':'#499BC0',
    'offshore':'#8FDEE3',
    'upv':'#F78779',
    'dpv':'#FDD786'
}

fig,ax = plt.subplots(figsize=(12,16),dpi=600)

ax.plot(vre_profit['inte_cum'],
        vre_profit['profit_cum'],
        color='pink',
        lw=1.5)

for i,r in vre_profit.iterrows():
    ax.bar([r['inte_cum']],
           [r['profit_cum']],
           width=r['inte'],
           color=color[r['vre']])

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.bar([0],[0],facecolor='#499BC0',label='Onshore wind')
ax.bar([0],[0],color='#8FDEE3',label='Offshore wind')
ax.bar([0],[0],color='#F78779',label='UPV')
ax.bar([0],[0],color='#FDD786',label='DPV')

ax.axhline(y=0,color='#9c5ef0',ls='-.',label='Zero line')

ax.set_xlim(0,41)
ax.set_ylim(-250,3000)

ax.minorticks_on()

ax.legend(reverse=True,
          edgecolor='black',
          prop={'size':22})

ax.set_xlabel('Cummulative integrated electricity (PWh) ')
ax.set_ylabel('Cummulative profit (100 M$)')


plt.savefig('./fig/vre_profit.jpg',bbox_inches='tight')
#plt.savefig('./fig/vre_profit.pdf',bbox_inches='tight')

plt.close()
```
