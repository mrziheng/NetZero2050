```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

# Figure6 a is plotted using arcgis

```python
emis_sp = pd.read_csv('./data/emis/emis_sp.csv',index_col=['region'])
emis_text = emis_sp.loc[emis_sp['emis_sp']>150]
```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 25

fig,ax = plt.subplots(figsize=(12.5,10.5),dpi=600)

cmap = sns.cubehelix_palette(rot=-.2, as_cmap=True)

sns.scatterplot(data=emis_sp,
                ax=ax,
                x="heat", 
                y="carbon_poten",
                hue="Cost\n[$/tCO2]", 
                size="Demand\n[TWh/yr]",
                palette='viridis_r', 
                sizes=(100, 1750))

for i,r in emis_text.iterrows():
    if i == 'UnitedArabEmirates':
        txt = 'United\nArabEmirates'
    elif i == 'Taiwan':
        txt = ''
    else:
        txt = i
    if i == 'UnitedArabEmirates':
        ax.text(r['heat']*4.5, 
                r['carbon_poten']*1.125, 
                txt, 
                fontsize=15, 
                ha='right', 
                va='bottom')
    else:
        ax.text(r['heat']*1.8, 
                r['carbon_poten']*1.125, 
                txt, 
                fontsize=15, 
                ha='right', 
                va='bottom')

ax.text(emis_sp.at['China','heat']*1.7,
        emis_sp.at['China','carbon_poten']*1.0, 
        'China', 
        fontsize=15, 
        ha='right', 
        va='bottom')

ax.text(emis_sp.at['USA','heat']*1.6,
        emis_sp.at['USA','carbon_poten']*1.0, 
        'USA', 
        fontsize=15, 
        ha='right', 
        va='bottom')

ax.axhline(100,ls=':',color='black')
ax.axvline(1,ls=':',color='black')

ax.set_xscale('log')  # 设置为对数坐标轴
ax.set_yscale('log')  # 设置为对数坐标轴

ax.set_xlabel('Biomass potential (EJ/yr)')
ax.set_ylabel('Carbon sequestration potential (Gt)')

ax.xaxis.grid(True, "minor", linewidth=.25)
ax.yaxis.grid(True, "minor", linewidth=.25)

ax.legend(edgecolor='white',ncol=1,prop={'size':15})

plt.xticks(font={'size':15})
plt.yticks(font={'size':15})

plt.savefig('./fig/emis_sp.jpg',bbox_inches='tight')
plt.savefig('./fig/emis_sp.pdf',bbox_inches='tight')

plt.close()
```

```python
emis = pd.read_csv('./data/emis/emis.csv',index_col=['scenario'])
capt = pd.read_csv('./data/emis/capt.csv',index_col=['scenario'])
```

```python

plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 20

fig,ax = plt.subplots(figsize=(12,11.5),dpi=600)

width = 0.5

color = {
    'Coal':'silver',
    'Gas':'#B288F3',
    'BECCS':"#30A865",
    'DACCS':"#417AD9",
    'Coal CCS':'silver',
    'Gas CCS':'#B288F3'
}

emis.plot(ax=ax,
          kind='bar',
          stacked=True,
          y=['coal','gas'],
          label=['Coal','Gas'],
          color=color,
          width=width,
          edgecolor='black')

capt.plot(ax=ax,
          kind='bar',
          stacked=True,
          y=['coal','gas','bio','dac'],
          label=['Coal CCS','Gas CCS','BECCS','DACCS'],
          width=width,
          color=color,
          hatch='///')

circle_style = dict(boxstyle='circle', facecolor='none', edgecolor='black')

for idx,scen in enumerate(emis.index):
    ax.text(idx,
            emis.at[scen,'tot']+200,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=15)

ax.axhline(0,color='black')

ax.set_ylim(-4000,4000)

ax.set_xticks(range(emis.shape[0]),['' for i in range(emis.shape[0])])

ax.set_xlabel('Scenarios',font={'size':25})
ax.set_ylabel('Carbon emission (MtCO$_2$)')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=6, 
               color='black')

ax.tick_params(axis='x', 
               which='major', 
               length=6, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='black')

dac_ax = fig.add_axes([0.22,0.18,0.285,0.145])

dac_ax.bar(range(capt.shape[0]),
           capt['dac'],
           color='#417AD9',
           hatch='///')

for idx,scen in enumerate(capt.index):
    dac_ax.text(idx,
            capt.at[scen,'dac']-4,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=10)

dac_ax.set_title('CO$_2$ captured by DACCS',font={'size':18})

dac_ax.set_xticks([],[])

dac_ax.set_ylim(-40,0)

# dac_ax.axhline(0,color='black')

dac_ax.set_ylabel('[MtCO$_2$]')

plt.savefig('./fig/emis_capt.jpg',bbox_inches='tight')
plt.savefig('./fig/emis_capt.pdf',bbox_inches='tight')

plt.close()
```
