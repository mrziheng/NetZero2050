```python
global_vis_day = pd.read_csv('./data/loadprof/global_vis_day.csv')
```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 25

power_col = ['Coal',
             'Gas',
             'Biomass',
             'Nuclear',
             'Hydropower',
             'Wind',
             'Solar PV']

power_color = {
    'Storage Discharge':'silver',
    'Trunk trans':'crimson',
    'Coal':'#2F322C',
    'Hydropower':'#2d768f',
    'Nuclear':'#6a3476',
    'Gas':'#334991',
    'Biomass':'#81c07a',
    'Solar PV':'#e46b50',
    'DPV':'#efce88',
    'Wind':'blueviolet',
    'Offshore wind':'#9DC1D9',
    'PHS':'royalblue',
    'BAT':'deepskyblue',
    'CAES':'coral',
    'VRB':'sienna',
    'CCS': 'palegreen',
    'Storage':'royalblue',
    'Reserve':'lightsteelblue'}


fig,ax = plt.subplots(figsize=(12.5,16),dpi=600)

global_vis_day.plot(ax=ax,
                     kind='area',
                     y=power_col,
                     stacked=True,
                     color=power_color,
                     legend=False,
                     lw=0.01)


global_vis_day.plot(ax=ax,
                     kind='line',
                     y="Demand",
                     color='black',
                     legend=False)


ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)

ax.set_ylabel("Load (TW)")
ax.set_xlabel("Time (Day)")

ax.legend(bbox_to_anchor=(0.5,1.01),
          borderaxespad=0,
          reverse=True,
          ncol=3,
          loc='upper center',
          frameon=False,
          prop={'size':20})

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=5, 
               color='black')

ax.set_ylim(0,200)
ax.set_xlim(0,365)

ax.minorticks_on()

ax.set_xticks(range(0,361,20),range(0,361,20),rotation=90)

ax.tick_params(axis='x', 
               which='major', 
               length=5, 
               color='black')

plt.savefig('./fig/load_prof_day.jpg',bbox_inches='tight')
plt.savefig('./fig/load_prof_day.pdf',bbox_inches='tight')

plt.close()
```
