```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Cairo')
```

```python
therm_cap = pd.read_csv('./data/Thermal/firm_cap_by_gen.csv')

therm_gen = pd.read_csv('./data/Thermal/thermal_gen.csv')
```

```python
therm_gen[['coal','nuclear','gas','bio']] = 1e-6 * therm_gen[['coal','nuclear','gas','bio']]
```

```python
therm_cap['sort_idx'] = therm_cap[['coal','gas','bio','nuclear']].sum(axis=1)

therm_gen['sort_idx'] = therm_cap['sort_idx']
```

```python
therm_cap = therm_cap.groupby(by='region').sum(numeric_only=1)
therm_gen = therm_gen.groupby(by='region').sum(numeric_only=1)
```

```python
therm_cap = therm_cap.sort_values(by='sort_idx',ascending=False)

therm_gen = therm_gen.sort_values(by='sort_idx',ascending=False)
```

```python

label = ['Coal','Nuclear','Gas','Biomass']

power_color = {
    'Storage Discharge':'silver',
    'Trunk trans':'crimson',
    'Coal':'#2F322C',
    'Hydropower':'#2d768f',
    'Nuclear':'#6a3476',
    'Gas':'#334991',
    'Biomass':'#81c07a',
    'Solar PV':'#e46b50',
    'DPV':'#efce88',
    'Wind':'#5B76B9',
    'Offshore wind':'#9DC1D9',
    'PHS':'royalblue',
    'BAT':'deepskyblue',
    'CAES':'coral',
    'VRB':'sienna',
    'CCS': 'palegreen',
    'Storage':'royalblue',
    'Reserve':'lightsteelblue'}

```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 25

width = 0.4

crs = 'epsg:4326'

fig,ax = plt.subplots(figsize=(16,10),dpi=600)

ax_twinx = ax.twinx()


therm_cap.loc[therm_cap['sort_idx']>25].plot(ax=ax,
                                             kind='bar',
                                             y=['coal','nuclear','gas','bio'],
                                             color=power_color,
                                             label=label,
                                             position=1,
                                             width=width,
                                             stacked=True)

therm_gen.loc[therm_gen['sort_idx']>25].plot(ax=ax_twinx,
                                             kind='bar',
                                             y=['coal','nuclear','gas','bio'],
                                             color=power_color,
                                             label=label,
                                             position=0,
                                             width=width,
                                             stacked=True,
                                             hatch='/',
                                             edgecolor='silver',
                                             alpha=0.75)


ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='black')

ax_twinx.minorticks_on()

ax_twinx.tick_params(axis='y', 
                     which='major', 
                     length=8, 
                     color='black')

ax_twinx.tick_params(axis='y', 
                     which='minor', 
                     length=4, 
                     color='black')

ax_twinx.tick_params(axis='x', 
                     which='minor', 
                     length=0, 
                     color='black')

ax.spines['top'].set_visible(False)

ax_twinx.spines['top'].set_visible(False)


#ax.grid(axis='x', alpha=0.5)

ax.set_xlabel('')
ax_twinx.set_xlabel('')

ax.set_ylabel('Installed Capacity (GW)')
ax_twinx.set_ylabel('Generation (PWh/yr)')

ax.set_xlim(-1,)
ax.set_ylim(0,1400)

ax.legend(reverse=True,
          loc='upper center',
          title='Capacity (left)',
          edgecolor='black')

ax_twinx.legend(reverse=True,
                loc='upper right',
                title='Generation (right)',
                edgecolor='black')

plt.savefig('./fig/therm_cap.jpg',bbox_inches='tight')

#plt.savefig('./fig/therm_cap.pdf',bbox_inches='tight')

plt.close()
```
