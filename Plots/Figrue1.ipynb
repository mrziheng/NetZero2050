```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
obj_struc = pd.read_csv('./data/power_struc/scoe_struc.csv')
power_struc = pd.read_csv('./data/power_struc/power_struc.csv',index_col=['scen'])
trans_dist = pd.read_csv('./data/power_struc/trans_cap_dist.csv',index_col=['scen'])
ele_struc = pd.read_csv('./data/power_struc/ele_struc.csv',index_col=['scen'])
dem_tot = pd.read_csv('./data/power_struc/demtot.csv')
```

```python
for k in ele_struc:
    ele_struc[k] = 1e-3 * ele_struc[k]

dem_tot['demand'] = 1e-3 * dem_tot['demand']
    
for k in power_struc:
    power_struc[k] = 1e-3 * power_struc[k]

for k in trans_dist:
    trans_dist[k] = 1e-3 * trans_dist[k]
```

```python
col_tag = {'Coal':['coal_fixed','coal_var',
                    'coal_starton','coal_closedown',
                    'coal_fuel','coal_rampup'],
            'Nuclear':['nuclear_fixed','nuclear_var',
                       'nuclear_starton','nuclear_closedown',
                       'nuclear_fuel','nuclear_rampup'],
            'Gas':['gas_fixed','gas_var','gas_starton',
                   'gas_closedown','gas_fuel','gas_rampup'],
            'Hydro':['hydro'],
            'Bio':['bio_fixed','bio_var','bio_starton',
                   'bio_closedown','bio_fuel','bio_rampup'],
            'CCS':["ccs_trans","ccs_sto","ccs_capture"],
            'Onshore wind':['onshore'],
            'Offshore wind':['offshore'],
            'UPV':['upv'],
            'DPV':['dpv'],
            'Storage':['phs','bat'],
            'UHV trans':['uhv_trans'],
            'Trunk trans':['trunk','hp_trunk'],
            'Spur trans':['spur','hp_spur']
        }
        
power_color = {
    'UHV trans':'purple',
    'Spur trans':'palevioletred',
    'Trunk trans':'crimson',
    'Coal':'#2F322C',
    'Hydro':'#2d768f',
    'Nuclear':'#6a3476',
    'Gas':'#334991',
    'Bio':'#81c07a',
    'UPV':'#e46b50',
    'DPV':'#efce88',
    'Onshore wind':'#5B76B9',
    'Offshore wind':'#9DC1D9',
    'PHS':'royalblue',
    'BAT':'deepskyblue',
    'CAES':'coral',
    'VRB':'sienna',
    'CCS': 'palegreen',
    'Storage':'royalblue',
    'Reserve':'lightsteelblue'}

power_col = ['coal','nuclear','gas','hydro','bio',
             'onshore','offshore','upv','dpv']
            
power_label = ['Coal','Nuclear','Gas','Hydro','Bio',
               'Onshore wind','Offshore wind','UPV','DPV']

sto_col = ['phs','bat']
sto_label = ['PHS','BAT']
        
col = list(col_tag.keys())
```

```python
plt.rcParams['font.size'] = 30
plt.rcParams['font.family'] = 'Arial'
    
width = 0.6
    
circle_style = dict(boxstyle='circle', facecolor='none', edgecolor='black')    
    
fig,axes = plt.subplots(3,3,figsize=(42,40),dpi=600,sharex=True)

for i in range(2):
        for j in range(2):
                axes[i][j].axis('off')

ax = axes[1][1]

ax.axis('off')

ax.bar([0],[0],facecolor='#2F322C',label='Coal')
ax.bar([0],[0],color='#6a3476',label='Nuclear')
ax.bar([0],[0],color='#334991',label='Gas')
ax.bar([0],[0],color='#2d768f',label='Hydropower')
ax.bar([0],[0],color='#81c07a',label='Biomass')
ax.bar([0],[0],color='palegreen',label='CCS cost')
ax.bar([0],[0],color='#5B76B9',label='Onshore wind')
ax.bar([0],[0],color='#9DC1D9',label='Offshore wind')
ax.bar([0],[0],color='#e46b50',label='UPV')
ax.bar([0],[0],color='#efce88',label='DPV')
ax.bar([0],[0],color='royalblue',label='Storage cost')
ax.bar([0],[0],color='purple',label='UHV cost')
ax.bar([0],[0],color='crimson',label='Trunk line cost')
ax.bar([0],[0],color='palevioletred',label='Spur line cost')
ax.bar([0],[0],facecolor='none',edgecolor='teal',label='Electricity demand')
        

ax.set_ylim(1,)

ax.legend(loc='center right',
          reverse=True,
          edgecolor='none',
          prop={'size':30},
          ncol=1)


ax = axes[0][2]

obj_struc.plot.bar(ax=ax,
                   y=col,
                   color=power_color,
                   stacked=True,
                   legend=False,
                   width=width)

obj_struc['tot'] = obj_struc[col].sum(axis=1)

for idx,scen in enumerate(obj_struc.index):
    ax.text(idx,
            obj_struc.at[scen,'tot']*1.075,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=25)

ax.text(-0.125,
        1.05,
        'b',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')


ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='blue')

ax.set_ylim(0,70)

ax.set_ylabel('System cost of electricity ($/MWh)')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


ax = axes[1][2]

ele_struc.plot.bar(ax=ax,
                   y=power_col,
                   stacked=True,
                   xlabel='',
                   label=power_label,
                   color=power_color,
                   legend=False,
                   width=width)

dem_tot.plot.bar(ax=ax,
                 y='demand',
                 label='Demand',
                 facecolor='none',
                 edgecolor='teal',
                 width=width,
                 legend=False,
                 linewidth=2.5)


ele_struc['tot'] = ele_struc[power_col].sum(axis=1)

for idx,scen in enumerate(ele_struc.index):
    ax.text(idx,
            ele_struc.at[scen,'tot']*1.05,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=25)

ax.set_ylabel('Electricity (PWh/yr)')

ax.text(-0.125,
        1.05,
        'c',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='blue')

ax.set_ylim(0,70)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


ax = axes[2][0]

power_struc.plot.bar(ax=ax,
                     y=power_col,
                     xlabel='',
                     stacked=True,
                     label=power_label,
                     color=power_color,
                     lw=0.1,
                     legend=False)

power_struc['tot'] = power_struc[power_col].sum(axis=1)

for idx,scen in enumerate(power_struc.index):
    ax.text(idx,
            power_struc.at[scen,'tot']*1.075,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=25)

ax.text(-0.125,
        1.05,
        'd',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='blue')

ax.set_xticks([],[])

ax.set_ylim(0,30)

ax.set_xticks(range(power_struc.shape[0]),power_struc.index)

ax.set_ylabel('Power capacity (TW)')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax = axes[2][1]

power_struc.plot.bar(ax=ax,
                     y=sto_col,
                     xlabel='',
                     stacked=True,
                     label=sto_label,
                     color=power_color,
                     lw=0.1)


power_struc['sto_tot'] = power_struc[sto_col].sum(axis=1)

for idx,scen in enumerate(power_struc.index):
    ax.text(idx,
            power_struc.at[scen,'sto_tot']+0.25,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=25)
    
ax.text(-0.125,
        1.05,
        'e',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='blue')
ax.set_xticks([],[])

ax.set_ylim(0,7)


ax.legend(frameon=False,reverse=True,prop={'size':20})

ax.set_ylabel('Storage capacity (TW)')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax = axes[2][2]

trans_dist.plot.bar(ax=ax,
                     y=['cap_dist'],
                     xlabel='',
                     color='purple',
                     lw=0.1,
                     legend=False,
                     alpha=0.5)

for idx,scen in enumerate(trans_dist.index):
    ax.text(idx,
            trans_dist.at[scen,'cap_dist']*1.075,
            str(idx+1),
            ha='center', 
            va='center',
            bbox=circle_style, 
            fontsize=25)

ax.text(-0.125,
        1.05,
        'f',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')

ax.minorticks_on()

ax.tick_params(axis='y', 
               which='major', 
               length=8, 
               color='black')

ax.tick_params(axis='y', 
               which='minor', 
               length=4, 
               color='black')

ax.tick_params(axis='x', 
               which='minor', 
               length=0, 
               color='blue')

ax.set_xticks([],[])


ax.set_ylim(0,8)

ax.set_ylabel('Transmission capacity (1000 TW-km)')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)


plt.subplots_adjust(hspace=0.09)

plt.savefig('./fig/power_struc.jpg',bbox_inches='tight')
plt.savefig('./fig/power_struc.pdf',bbox_inches='tight')

plt.close()
```
