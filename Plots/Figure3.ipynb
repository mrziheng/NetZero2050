```python
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Polygon
import pandas as pd
```

# Figure 5 a1 and a2 is plotted using arcgis

```python
world = gpd.read_file('./data/shp/territory_vis.shp')
```

```python
def add_colorbar(fig,ax,vmax,label,vmin=0,cmap='autumn_r',loc=[0.175,0.39,0.02,0.2],extend='max'):
        
    cax = ax.inset_axes(loc,transform=ax.transAxes)
    
    im = plt.cm.ScalarMappable(cmap=cmap,
                               norm=plt.Normalize(vmin=vmin,
                                                  vmax=vmax))
    
    if extend == 'none':
        cbar = fig.colorbar(im,cax=cax)
    else:
        cbar = fig.colorbar(im,cax=cax,extend=extend)
    
    cbar.outline.set_edgecolor('white')
    
    cbar.dividers.set_color('red')
    
    cbar.ax.tick_params(labelsize=15) 
    
    cbar.minorticks_on()
    
    cax.yaxis.tick_left()
    
    cbar.set_label(label,font={'size':25})
    


def draw_attr_map(fig,ax,geo,col,cmap,vmax,cbar_label,cbar_loc,draw_edge=False,extend='max'):
    
    world.plot(ax=ax,
               facecolor='whitesmoke',
               edgecolor='whitesmoke',
               lw=0.8)
    
    if draw_edge:
        geo.plot(ax=ax,
                column=col,
                cmap=cmap,
                vmax=vmax,
                edgecolor='silver',
                linewidth=0.25)
    else:
        geo.plot(ax=ax,
                column=col,
                cmap=cmap,
                vmax=vmax)
        
    
    add_colorbar(fig=fig,
                 ax=ax,
                 label=cbar_label,
                 vmax=vmax,
                 cmap=cmap,
                 loc=cbar_loc,
                 extend=extend)
```

```python
upv_res = pd.read_csv('./data/vre_res/upv_attr_era5_cell_2019.csv')
dpv_res = pd.read_csv('./data/vre_res/dpv_attr_era5_cell_2019.csv')
onshore_res = pd.read_csv('./data/vre_res/onshore_attr_era5_cell_2019.csv')
offshore_res = pd.read_csv('./data/vre_res/offshore_attr_era5_cell_2019.csv')
```

```python
geo = []

for i,r in upv_res.iterrows():
    geo.append(Polygon([(r['lon'],r['lat']),
                        (r['lon']+0.25,r['lat']),
                        (r['lon']+0.25,r['lat']-0.25),
                        (r['lon'],r['lat']-0.25)]))
    
upv_res_gdf = gpd.GeoDataFrame(upv_res,geometry=geo,crs='epsg:4326')
```

```python
geo = []

for i,r in dpv_res.iterrows():
    geo.append(Polygon([(r['lon'],r['lat']),
                        (r['lon']+0.25,r['lat']),
                        (r['lon']+0.25,r['lat']-0.25),
                        (r['lon'],r['lat']-0.25)]))
    
dpv_res_gdf = gpd.GeoDataFrame(dpv_res,geometry=geo,crs='epsg:4326')
```

```python
we_res = pd.concat([onshore_res,offshore_res])

geo = []

for i,r in we_res.iterrows():
    geo.append(Polygon([(r['lon'],r['lat']),
                        (r['lon']+0.25,r['lat']),
                        (r['lon']+0.25,r['lat']-0.25),
                        (r['lon'],r['lat']-0.25)]))
    
we_res_gdf = gpd.GeoDataFrame(we_res,geometry=geo,crs='epsg:4326')
```

```python
vis_upv_res_gdf = upv_res_gdf.loc[upv_res_gdf['if']>0]
vis_dpv_res_gdf = dpv_res_gdf.loc[dpv_res_gdf['if']>0]
vis_we_res_gdf = we_res_gdf.loc[we_res_gdf['if']>0]
```

```python
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 18


fig,ax = plt.subplots(figsize=(24,10),dpi=800)


draw_attr_map(fig=fig,
              ax=ax,
              geo=vis_upv_res_gdf,
              col='if',
              cmap='autumn_r',
              cbar_label='Installation factor (0~1)',
              vmax=1.0,
              cbar_loc=[0.05,0.275,0.02,0.4],
              extend='none')

ax.text(0.015,
        0.95,
        'a',
        ha='center',
        va='center',
        transform=ax.transAxes,
        size = 35,
        weight='bold')

ax.set_xticks([],[])
ax.set_yticks([],[])

plt.savefig('./fig/upv_if.jpg',bbox_inches='tight')
plt.savefig('./fig/upv_if.pdf',bbox_inches='tight')

plt.close()
```
